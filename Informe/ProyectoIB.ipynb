### OBJETIVOS

* Modelar la infraestructura del campues como un grafo no dirigido donde los nodos representan edificios y las aristas representan conexiones transitables.
* Implementar el algoritmo BFS para determinar la ruta más corta entre un punto de origen y un punto de evacuación. 
* Implementar el algoritmo DFS para explorar todas las posibles rutas y encontrar el camino más profundo disponible. 

### INTRODUCCION

los sistemas de evacuación en entornos complejos como campus universitarios requieren herramientas que permiten calcular rutas óptimas de manera rápida y confiable. En este proyecto se implementa dos algoritmos fundamentales de teoría de grafos aplicados al problema de evacuación como lo son BFS y DFS. 

#### Generacion

El campus se modela mediante un grafo no dirigido donde:

* Nodos (Vértices) : Representan edificios del campues (CEC, FIQA, Biblioteca, etc).
* Aristas (Conexiones): Representan caminos transitables entre edificios.

La estructura del grafo se almacena en formato csv con pares de edificios conectados. El archivo "CampusEpn.csv" contiene 16 conexiones entre edificios como: 

CEC ↔ Salida1

ESFOT ↔ FIS

Electrica ↔ Mecanica

Civil ↔ CIENCIAS

La clase "Grafo" implementa la estructura de datos mediante: 

* Un diccionario de nodos (self.nodos)

* Listas de adyacencia para cada nodo

* Métodos para agregar/eliminar nodos y aristas

#### BFS 

El algoritmo BFS (Búsqueda en Anchura) explora el grafo nivel por nivel utilizando una cola (estructura FIFO). Garantiza encontrar la ruta más corta en grafos no ponderados.

Características principales:

* Explora todos los vecinos de un nodo antes de pasar al siguiente nivel
  
* Utiliza una cola (deque) para gestionar los nodos por visitar
  
* Registra distancias desde el nodo origen

* Construye un árbol de padres que permite reconstruir la ruta

Aplicación en evacuación:

Encuentra la salida más cercana (menor número de edificios que atravesar), también es ideal para evacuaciones urgentes donde se prioriza velocidad, además proporciona la ruta con menor número de decisiones

#### DFS 

El algoritmo DFS (Búsqueda en Profundidad) explora el grafo siguiendo un camino hasta su máxima profundidad antes de retroceder. Utiliza recursión o una pila (estructura LIFO).

Características principales:

* Explora cada rama completamente antes de probar otras
  
* Implementación recursiva con backtracking
  
* Encuentra múltiples rutas posibles
  
* Identifica el camino más largo (más profundo)

Aplicación en evacuación:

Es útil para análisis de cobertura de rutas alternativas, también permite identificar rutas de respaldo y encuentra caminos menos obvios que podrían estar menos congestionados

### DESARROLLO

#### Implementacion

*Estructura de Clases*

    Clase Nodo: 

```python
nombre: identificador del edificio
visitado: bandera para control de recorrido
adyacentes: lista de nodos conectados
```

    Clase Grafo: 

```python
nodos: diccionario {nombre: objeto_nodo}
agregar_nodo(nombre): añade un nuevo edificio
agregar_arista(origen, destino): conecta dos edificios bidireccionalmente
eliminar_arista(origen, destino): elimina conexión
```

**Algoritmo BFS**

La función "bfs(grafo, inicio, fin)" retorna un diccionario con:

* distancias: distancia mínima desde inicio a cada nodo alcanzable

* padres: árbol de exploración BFS

* ruta_mas_corta: secuencia de edificios de la ruta óptima

* tiempo: tiempo de ejecución en segundos

**Proceso:**

1. Inicializa cola con nodo inicial

2. Marca inicio como visitado con distancia 0

3. Mientras la cola no esté vacía:

* Extrae nodo actual

* Si es el destino, reconstruye y retorna la ruta

* Para cada vecino no visitado: marca como visitado, registra padre y distancia, añade a la cola



**Algoritmo DFS**

La función "dfs(grafo, inicio, fin)" retorna un diccionario con:

* padres: árbol de exploración DFS

* ruta_encontrada: primera ruta encontrada por DFS

* camino_mas_profundo: ruta más larga entre todas las encontradas

* tiempo: tiempo de ejecución en segundos

**Proceso:**

1. Utiliza recursión con backtracking

2. Explora cada vecino no visitado recursivamente

3. Registra todas las rutas encontradas

4. Retorna la primera ruta (simulando comportamiento DFS estándar) y la más profunda

### COMPLEJIDAD 

*Complejidad Temporal*

**BFS:**

* Peor caso: O(V + E)

V = número de vértices (edificios)

E = número de aristas (conexiones)


* Cada nodo se visita exactamente una vez
  
* Cada arista se examina exactamente una vez

**DFS (implementación con todas las rutas)**:

* Peor caso: O(V!) en grafos densos

* Explora todas las rutas posibles mediante backtracking

* En el grafo del campus: O(V + E) para construcción del árbol estándar

* Búsqueda de todas las rutas: exponencial en el peor caso

*Complejidad Espacial*

**BFS:**

O(V) para la cola y estructuras auxiliares (visitados, padres, distancias)

En el peor caso, la cola puede contener todos los nodos de un nivel

**DFS:**

O(V) para la pila de recursión (en el camino más profundo)

O(V×R) para almacenar todas las rutas encontradas (R = número de rutas)

**Análisis para el Campus EPN**

Con V = 13 nodos y E = 16 aristas:

* BFS: O(29) operaciones → muy eficiente

* DFS: Depende de la conectividad, pero manejable para este tamaño

### CONCLUSIONES

* BFS es superior para encontrar rutas de evacuación óptimas en términos de distancia, garantizando el camino más corto en grafos no ponderados.

* DFS proporciona información valiosa sobre rutas alternativas y cobertura del espacio de búsqueda, útil para planificación de contingencias.
  
* La interfaz gráfica desarrollada facilita la visualización de rutas, permitiendo a administradores del campus tomar decisiones informadas sobre evacuación.

* La representación del campus como grafo no dirigido es efectiva y escalable, permitiendo añadir nuevos edificios y conexiones dinámicamente.
