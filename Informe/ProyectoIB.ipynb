### OBJETIVOS

* Modelar la infraestructura del campues como un grafo no dirigido donde los nodos representan edificios y las aristas representan conexiones transitables.
* Implementar el algoritmo BFS para determinar la ruta más corta entre un punto de origen y un punto de evacuación. 
* Implementar el algoritmo DFS para explorar todas las posibles rutas y encontrar el camino más profundo disponible. 

### INTRODUCCION

los sistemas de evacuación en entornos complejos como campus universitarios requieren herramientas que permiten calcular rutas óptimas de manera rápida y confiable. En este proyecto se implementa dos algoritmos fundamentales de teoría de grafos aplicados al problema de evacuación como lo son BFS y DFS. 

#### Generacion

El campus se modela mediante un grafo no dirigido donde:

* Nodos (Vértices) : Representan edificios del campues (CEC, FIQA, Biblioteca, etc).
* Aristas (Conexiones): Representan caminos transitables entre edificios.

La estructura del grafo se almacena en formato csv con pares de edificios conectados. El archivo "CampusEpn.csv" contiene 16 conexiones entre edificios como: 

CEC ↔ Salida1

ESFOT ↔ FIS

Electrica ↔ Mecanica

Civil ↔ CIENCIAS

La clase "Grafo" implementa la estructura de datos mediante: 

* Un diccionario de nodos (self.nodos)

* Listas de adyacencia para cada nodo

* Métodos para agregar/eliminar nodos y aristas
